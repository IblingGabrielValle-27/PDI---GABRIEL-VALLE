```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

def blur_2(image, mask):
    fondo = cv2.GaussianBlur(image, (51, 51), 0)
    result = np.where(mask[:, :, None] > 0, image, fondo)
    return result

image = cv2.imread('rosa.png')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
_, mask = cv2.threshold(gray, 100, 255, cv2.THRESH_BINARY)
mask = cv2.bitwise_not(mask)
mask = cv2.GaussianBlur(mask, (21, 21), 0)
resultado = blur_2(image, mask)

plt.figure(figsize=(10, 5))
plt.subplot(121)
plt.title("Original")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(122)
plt.title("Desenfoque selectivo")
plt.imshow(cv2.cvtColor(resultado, cv2.COLOR_BGR2RGB))
plt.axis('off')
plt.show()
```

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt

image = cv2.imread('elementos.jpg')
gray = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)
_, binario = cv2.threshold(gray, 128, 255, cv2.THRESH_BINARY_INV + cv2.THRESH_OTSU)
contornos, _ = cv2.findContours(binario, cv2.RETR_EXTERNAL, cv2.CHAIN_APPROX_SIMPLE)

imagen_con_bordes = image.copy()
cv2.drawContours(imagen_con_bordes, contornos, -1, (0, 255, 0), 2)

cont = len(contornos)
print(f"Número de elementos detectados: {cont}")

plt.figure(figsize=(12, 6))
plt.subplot(121)
plt.title("Original")
plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.subplot(122)
plt.title("Contornos detectados")
plt.imshow(cv2.cvtColor(imagen_con_bordes, cv2.COLOR_BGR2RGB))
plt.axis('off')

plt.tight_layout()
plt.show()
```
