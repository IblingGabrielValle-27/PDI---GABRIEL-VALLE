Realiza un pequeño script que permite visualizar una lista de paleta de colores disponibles en OPENCV a través de un ciclo repetitivo

REFERENCIAS
1. https://docs.opencv.org/4.x/d3/d50/group__imgproc__colormap.html#gga9a805d8262bcbe273f16be9ea2055a65a5d4a709cda1130bfa3cc4b4c5ae8f82a

2. https://omes-va.com/deteccion-de-colores/

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('2.png', 0)

colormap_list = [
    (cv2.COLORMAP_JET, 'JET'),
    (cv2.COLORMAP_HSV, 'HSV'),
    (cv2.COLORMAP_OCEAN, 'OCEAN'),
    (cv2.COLORMAP_PINK, 'PINK'),
    (cv2.COLORMAP_SPRING, 'SPRING'),
    (cv2.COLORMAP_COOL, 'COOL'),
    (cv2.COLORMAP_WINTER, 'WINTER'),
    (cv2.COLORMAP_RAINBOW, 'RAINBOW'),
    (cv2.COLORMAP_BONE, 'BONE'),
    (cv2.COLORMAP_TWILIGHT, 'TWILIGHT')
]

plt.figure(figsize=(15, 8))
for i, (colormap, name) in enumerate(colormap_list):
    img_colored = cv2.applyColorMap(image, colormap)
    
    plt.subplot(2, 5, i + 1)
    plt.imshow(cv2.cvtColor(img_colored, cv2.COLOR_BGR2RGB))
    plt.title(f'{name}')
    plt.axis('off')

plt.tight_layout()
plt.show()
```
