```python
import numpy as np
import cv2

# Crear una imagen en escala de grises de 512x512 píxeles
height, width = 512, 512
image = np.zeros((height, width), dtype=np.uint8)

# Dibujar formas geométricas con bordes claros
# Fondo blanco
image[:, :] = 255

# Rectángulo negro en el centro
image[150:350, 150:350] = 0

# Círculo gris
center = (width // 2, height // 2)
radius = 80
cv2.circle(image, center, radius, 100, -1)  # Gris medio (100)

# Líneas diagonales
cv2.line(image, (100, 100), (400, 400), 50, 5)  # Línea gris oscuro
cv2.line(image, (100, 400), (400, 100), 50, 5)  # Línea gris oscuro

# Guardar la imagen
cv2.imwrite('imagen_ejemplo.jpg', image)
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Cargar imagen
imagen = cv2.imread('imagen_ejemplo.jpg', cv2.IMREAD_GRAYSCALE)

# Detección de bordes Laplaciano
laplaciano = cv2.Laplacian(imagen, cv2.CV_64F)
laplaciano = np.uint8(np.absolute(laplaciano))

# Laplaciano del Gaussiano (LoG)
# Aplicar filtro Gaussiano
gaussiano = cv2.GaussianBlur(imagen, (5, 5), 0)
# Aplicar Laplaciano
log = cv2.Laplacian(gaussiano, cv2.CV_64F)
log = np.uint8(np.absolute(log))

# Mostrar resultados
plt.figure(figsize=(15, 5))

plt.subplot(1, 3, 1)
plt.title('Imagen Original')
plt.imshow(imagen, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 2)
plt.title('Laplaciano')
plt.imshow(laplaciano, cmap='gray')
plt.axis('off')

plt.subplot(1, 3, 3)
plt.title('Laplaciano del Gaussiano')
plt.imshow(log, cmap='gray')
plt.axis('off')

plt.show()
```
