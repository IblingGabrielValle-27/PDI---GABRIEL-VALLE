CONVOLUCIÓN

CONVOLUCIÓN ENFOQUE

```python
import cv2
import numpy as np 
import matplotlib.pyplot as plt

img = cv2.imread('image.png')
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
mascara = np.array([[0, -1, 0], [-1, 5, -1], [0, -1, 0]])
img_fin = cv2.filter2D(img_rgb, -1, mascara)
plt.subplot(121)
plt.imshow(img_rgb)
plt.title('Imagen Original')
plt.subplot(122)
plt.imshow(img_fin)
plt.title('Convolución Enfoque')
plt.show()
```
