```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucionmediohorizontal(imagen,tamanioKernel):
    kernel= np.ones((1,tamanioKernel),np.float32)/tamanioKernel
    imgfin= cv2.filter2D(imagen,-1,kernel)
    return imgfin
imagen= cv2.imread('img.jpeg')
gris= cv2.cvtColor(imagen,cv2.COLOR_BGR2GRAY)

imgrest= convolucionmediohorizontal(gris,9)

plt.figure(figsize=(10,5))
plt.subplot(121)
plt.imshow(cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB))

plt.subplot(122)
plt.imshow(imgrest,cmap='gray')

plt.show
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def convolucionmediohorizontal(imagen,tamanioKernel):
    kernel= np.ones((1,tamanioKernel),np.float32)/tamanioKernel
    imgfin= cv2.filter2D(imagen,-1,kernel)
    return imgfin
imagen= cv2.imread('img.jpeg')
imgrgb= cv2.cvtColor(imagen,cv2.COLOR_BGR2RGB)

imgrestr= convolucionmediohorizontal(imgrgb[:,:,0],25)
imgrestg= convolucionmediohorizontal(imgrgb[:,:,1],25)
imgrestb= convolucionmediohorizontal(imgrgb[:,:,2],25)

imgcolor=np.stack([imgrestr,imgrestg,imgrestb],axis=2)

plt.figure(figsize=(10,5))
plt.subplot(121)
plt.imshow(imgrgb)

plt.subplot(122)
plt.imshow(imgcolor)

plt.show
```
