ALIASING Y PATRONES DE MOIRÉ

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

def reduce_aliasing(image_path):
    image = cv2.imread(image_path)
    image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)
    smoothed_image = cv2.GaussianBlur(image_rgb, (9, 9), 0)
    return image_rgb, smoothed_image

image_rgb, smoothed_image = reduce_aliasing('IMG1.jpeg')

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image_rgb)
plt.title("Imagen Original")
plt.axis('off')

plt.subplot(1, 2, 2)
plt.imshow(smoothed_image)
plt.title("Imagen Mejorada")
plt.axis('off')

plt.show()
```

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

image = cv2.imread('IMG2.jpeg')
image_rgb = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

filtered_image = cv2.bilateralFilter(image_rgb, d=9, sigmaColor=100, sigmaSpace=100)

plt.figure(figsize=(12, 6))

plt.subplot(121)
plt.imshow(image_rgb)
plt.title("Original con Aliasing")
plt.axis('off')

plt.subplot(122)
plt.imshow(filtered_image)
plt.title("Imagen Mejorada")
plt.axis('off')

plt.show()
```
