EJERCICIO 1: Recorta el punto de interés de la imagen1.jpg tal cual se muestra en la figura, guarda la imagen recortafa en tu disco local, y muestra la imagen original y la recortada con axis (imagen a colores)

```python
import cv2
import matplotlib.pyplot as plt

imagen = cv2.imread('imagen1.jpg')
imagen_rgb = cv2.cvtColor(imagen, cv2.COLOR_BGR2RGB)
imagen_cortada = imagen_rgb[590:850, 670:840, :]

plt.imsave('img_cortada.jpg', imagen_cortada)

plt.figure(figsize=(10, 5))
plt.subplot(1, 2, 1)
plt.title('Imagen original')
plt.imshow(imagen_rgb)

plt.subplot(1, 2, 2)
plt.title('Imagen Cortada')
plt.imshow(imagen_cortada)

plt.show()
```

EJERCICIO 2: Genera un degradado lineal a colores

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
from matplotlib.colors import LinearSegmentedColormap
#Definicion de colores
colores=[(0,'blue'),(0.5,'green'),(1,'yellow')] # para generar escala de grices
cmap=LinearSegmentedColormap.from_list('degradado',colores)#listar los colores y generara color map a traves del array
imagen_degradado=cmap(np.linspace(0,1,256).reshape(-1,1))
plt.imshow(imagen_degradado,aspect='auto')
plt.show()
```

EJERCICIO 3: Genera una imagen semejante a un tablero de ajedrez tomando en cuenta que cada casilla debe medir 100x100 px

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
tamano = 100
tablero = np.zeros((8*tamano, 8*tamano, 3), dtype=np.uint8)
for i in range(8):
    for j in range(8):
        if (i+j) % 2 == 0:
            tablero[i*tamano:(i+1)*tamano, j*tamano:(j+1)*tamano] = (255, 255, 255)

plt.imshow(tablero)
plt.title('Tablero de Ajedrez')
plt.show()
```

EJERCICIO 5: Tomando en cuenta la imagen1.jpg aplica un degradado lineal en ella

```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

imagen = cv2.imread('imagen1.jpg', 0)
rows, cols = imagen.shape
deg_imagen = np.tile(np.linspace(255, 0, cols, dtype=np.uint8), (rows, 1))
imagen_degradada = cv2.addWeighted(imagen, 0.5, deg_imagen, 0.5, 0)

plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(imagen, cmap='gray')
plt.title('Imagen Original')

plt.subplot(1, 2, 2)
plt.imshow(imagen_degradada, cmap='gray')
plt.title('Imagen con Degradado')

plt.show()
```
